Parte A — El problema

Contexto

El dataset seleccionado contiene información sobre diferentes vinos tintos a partir de sus características físico-químicas. Entre las variables disponibles se encuentran la acidez, el azúcar residual, los cloruros, la densidad, el pH, los sulfatos y el porcentaje de alcohol. Además, cada registro posee una valoración de calidad.

Este tipo de información puede ser utilizada para analizar qué características están relacionadas con diferentes niveles de calidad del vino.

Objetivo

El objetivo de este proyecto es analizar las características físico-químicas de los vinos para identificar relaciones frecuentes entre ellas y construir un modelo que permita clasificar los vinos según categorías de calidad.

El análisis busca generar conocimiento que pueda servir como apoyo para comprender qué características se presentan en conjunto y qué variables permiten diferenciar los niveles de calidad.

¿Por qué utilizar minería de datos?

Este problema puede abordarse mediante minería de datos porque existen varias características que pueden relacionarse entre sí. Un promedio o una consulta simple permitiría observar una variable de manera aislada, pero no permitiría descubrir fácilmente combinaciones de características ni construir reglas de clasificación.

Por este motivo, se utilizarán reglas de asociación para encontrar relaciones entre características y un árbol de decisión para identificar condiciones que permitan clasificar los registros según su categoría de calidad.
